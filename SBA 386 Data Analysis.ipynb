```python
# Task 7: Create a bar chart showing total profit for each month
```

```python
plt.bar(df['month_number'], df['total_profit'])

plt.xlabel('Month number')
plt.ylabel('Total profit')
plt.title('Company profit per month')

plt.show() 
```

```python
# Task 8: Create a red dotted line chart showing monthly total profit
```

```python
# Task 8: Create a red dotted line chart showing monthly total profit

plt.plot(
    df['month_number'],
    df['total_profit'],
    color='red',
    linestyle=':',
    marker='o',
    markerfacecolor='black',
    linewidth=3,
    label='Profit data of last year'
)

plt.xlabel('Month number')
plt.ylabel('Total profit')
plt.title('Company profit per month')
plt.legend(loc='lower right')

plt.show()
```

```python
# Task 9: Display sales data for all products using a multiline plot

plt.plot(df['month_number'], df['facecream'], label='Face Cream')
plt.plot(df['month_number'], df['facewash'], label='Face Wash')
plt.plot(df['month_number'], df['toothpaste'], label='Toothpaste')
plt.plot(df['month_number'], df['bathingsoap'], label='Bathing Soap')
plt.plot(df['month_number'], df['shampoo'], label='Shampoo')
plt.plot(df['month_number'], df['moisturizer'], label='Moisturizer')

plt.xlabel('Month number')
plt.ylabel('Sales units')
plt.title('Sales data')

plt.legend()
plt.show()
```

```python
# Section Two - Question 2: Plot height and weight data

height = [121.9, 124.5, 129.5, 134.6, 139.7, 147.3, 152.4, 157.5, 162.6]
weight = [19.7, 21.3, 23.5, 25.9, 28.5, 32.1, 35.7, 39.6, 43.2]

plt.plot(
    weight,
    height,
    color='green',
    linestyle='-.',
    marker='o',
    markersize=10,
    label='Average weight'
)

plt.xlabel('Weight in kg')
plt.ylabel('Height in cm')
plt.title('Average weight with respect to average height')
plt.legend(loc='lower right')

plt.show()
```
